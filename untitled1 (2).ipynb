```python
# 1. TẠO DATAFRAME

import pandas as pd
import matplotlib.pyplot as plt

# Tạo dữ liệu của 10 sinh viên
data = {
    "Ho_ten": [
        "Nguyen Van An", "Le Thi Binh", "Nguyen Ngoc Chi", "Vo Thi Ha ", "Le An Han",
        "Tran Thi Lan", "Nguyen Minh Anh", "Quach Bao Nam", "Nguyen Phuc An", "Le Bao Trang"
    ],
    "Chuyen_can": [8, 9, 7, 8, 9, 6, 8, 7, 8, 9],
    "Giua_ky":    [7, 8, 6, 9, 7, 6, 8, 7, 7, 9],
    "Cuoi_ky":    [8, 8, 6, 9, 7, 5, 9, 6, 7, 10]
}

# Chuyển dữ liệu thành bảng
df = pd.DataFrame(data)

# Hiển thị bảng dữ liệu
display(df)
```

```python
# 2. TÍNH ĐIỂM TỔNG KẾT

df["Tong_ket"] = (
    0.2 * df["Chuyen_can"]
    + 0.3 * df["Giua_ky"]
    + 0.5 * df["Cuoi_ky"]
)

# Làm tròn đến 2 chữ số thập phân
df["Tong_ket"] = df["Tong_ket"].round(2)

# Hiển thị bảng kết quả
display(df)
```

```python
# 3. XẾP LOẠI SINH VIÊN

def xep_loai(diem):
    if diem >= 8.5:
        return "Giỏi"
    elif diem >= 7.0:
        return "Khá"
    elif diem >= 5.0:
        return "Trung bình"
    else:
        return "Yếu"

# Tạo cột xếp loại
df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)

# Hiển thị bảng điểm đầy đủ
display(df)
```

```python
# 4. THỐNG KÊ

# 1. Điểm tổng kết trung bình
diem_trung_binh = df["Tong_ket"].mean()

# 2. Sinh viên có điểm cao nhất
diem_cao_nhat = df["Tong_ket"].max()
sv_cao_nhat = df[df["Tong_ket"] == diem_cao_nhat]

# 3. Sinh viên có điểm thấp nhất
diem_thap_nhat = df["Tong_ket"].min()
sv_thap_nhat = df[df["Tong_ket"] == diem_thap_nhat]

# 4. Đếm số sinh viên đạt từ 5 điểm trở lên
so_sinh_vien_dat = (df["Tong_ket"] >= 5).sum()

# Hiển thị kết quả
print("Điểm trung bình của lớp:", round(diem_trung_binh, 2))

print("\nSinh viên có điểm cao nhất:")
display(sv_cao_nhat[["Ho_ten", "Tong_ket"]])

print("\nSinh viên có điểm thấp nhất:")
display(sv_thap_nhat[["Ho_ten", "Tong_ket"]])

print("\nSố sinh viên đạt:", so_sinh_vien_dat)
```

```python
# 5. TRỰC QUAN HÓA DỮ LIỆU

plt.figure(figsize=(10, 5))

plt.bar(df["Ho_ten"], df["Tong_ket"])

plt.title("ĐIỂM TỔNG KẾT CỦA 10 SINH VIÊN")
plt.xlabel("Họ và tên sinh viên")
plt.ylabel("Điểm tổng kết")

plt.ylim(0, 10)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
